```python
### 1.2 Custom Dataset Class

import os
from torch.utils.data import Dataset
from PIL import Image

class CustomFood101(Dataset):
    def __init__(self, root, split='train', transform=None):
        self.root = root
        self.transform = transform
        classes_file = os.path.join(root, 'meta', 'classes.txt')
        with open(classes_file, 'r') as f:
            self.classes = [line.strip() for line in f.readlines()]
        self.class_to_idx = {cls: idx for idx, cls in enumerate(self.classes)}
        split_file = os.path.join(root, 'meta', f'{split}.txt')
        with open(split_file, 'r') as f:
            self.samples = [line.strip() for line in f.readlines()]
        print(f"Loaded {len(self.samples)} images for {split} split")
        print(f"Number of classes: {len(self.classes)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        sample = self.samples[idx]
        class_name, image_name = sample.split('/')
        img_path = os.path.join(self.root, 'images', class_name, f'{image_name}.jpg')
        image = Image.open(img_path).convert('RGB')
        label = self.class_to_idx[class_name]
        if self.transform:
            image = self.transform(image)
        return image, label

print("CustomFood101 ready")
```

# Data Transformation & Model Init

```python
### 1.3 Data Transformation & DataLoader

!pip install timm -q

import torch
import torch.nn as nn
import timm
from torchvision import transforms
from torch.utils.data import DataLoader, random_split, Subset
from torchvision.transforms import RandAugment
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

IMG_SIZE = 300

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandAugment(num_ops=2, magnitude=9),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# Define separate datasets to prevent train/val transform leakage
train_full = CustomFood101(
    root='/content/food101_local/food-101',
    split='train',
    transform=train_transform
)

val_full = CustomFood101(
    root='/content/food101_local/food-101',
    split='train',
    transform=val_transform
)

train_size = int(0.8 * len(train_full))
val_size = len(train_full) - train_size
torch.manual_seed(42)
train_indices, val_indices = random_split(range(len(train_full)), [train_size, val_size])

train_dataset = Subset(train_full, train_indices.indices)
val_dataset = Subset(val_full, val_indices.indices)

batch_size = 32
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True,
                          num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False,
                        num_workers=2, pin_memory=True)
print(f"{len(train_dataset):,} training, {len(val_dataset):,} validation")
print(f"Batch size: {batch_size}, Image size: {IMG_SIZE}")


### 1.4 Model Initialization & Loss Function

model = timm.create_model('efficientnet_b3', pretrained=True, num_classes=101)

in_features = model.classifier.in_features
model.classifier = nn.Sequential(
    nn.Dropout(0.4),
    nn.Linear(in_features, 101)
)
model = model.to(device)

criterion = nn.CrossEntropyLoss(label_smoothing=0.1)


# HELPER FUNCTIONS
def freeze_bn(model):
    for m in model.modules():
        if isinstance(m, nn.BatchNorm2d):
            m.eval()
            m.weight.requires_grad = False
            m.bias.requires_grad = False

def evaluate(model, loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, pred = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (pred == labels).sum().item()
    return 100 * correct / total
```

# EDA

```python
## 2. Exploratory Data Analysis (EDA)

import os
import random
from PIL import Image
from tqdm import tqdm

DATA_ROOT = '/content/food101_local/food-101'
IMAGES_DIR = os.path.join(DATA_ROOT, 'images')

with open(os.path.join(DATA_ROOT, 'meta', 'classes.txt'), 'r') as f:
    classes = [line.strip() for line in f.readlines()]

# Sample 5000
all_images = []
for cls in classes:
    cls_dir = os.path.join(IMAGES_DIR, cls)
    for img_name in os.listdir(cls_dir):
        all_images.append(os.path.join(cls_dir, img_name))

random.seed(42)
sample = random.sample(all_images, min(5000, len(all_images)))
print(f"Sample: {len(sample)} images")

# Check corrupt
corrupt = []
for img_path in tqdm(sample, desc="Check corrupt"):
    try:
        img = Image.open(img_path)
        img.verify()
    except Exception as e:
        corrupt.append((img_path, str(e)))

print(f"\nCorrupt: {len(corrupt)}")
if corrupt:
    for p, e in corrupt[:5]:
        print(f"  {p}: {e}")
else:
    print("No corrupt")
```

```python
import hashlib
from tqdm import tqdm

sample_dup = random.sample(all_images, min(5000, len(all_images)))
hashes = {}
duplicates = []

for img_path in tqdm(sample_dup, desc="Check duplicates"):
    try:
        with open(img_path, 'rb') as f:
            h = hashlib.md5(f.read()).hexdigest()
        if h in hashes:
            duplicates.append((img_path, hashes[h]))
        else:
            hashes[h] = img_path
    except:
        continue

print(f"Sample: {len(sample_dup)} images")
print(f"Duplicate: {len(duplicates)}")
if duplicates:
    for p1, p2 in duplicates[:5]:
        print(f"  {p1}")
        print(f"  {p2}")
else:
    print("No duplicate")
```

# Training

```python
## 3. Model Training Strategy

print("\nPHASE 1: Training classifier only...")

for param in model.parameters():
    param.requires_grad = False
for param in model.classifier.parameters():
    param.requires_grad = True

freeze_bn(model)

optimizer = optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-3, weight_decay=1e-4
)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=3)

best_acc = 0
for epoch in range(3):
    model.train()
    freeze_bn(model)
    correct, total, running_loss = 0, 0, 0
    for images, labels in tqdm(train_loader, desc=f"Phase 1 - Epoch {epoch+1}"):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * labels.size(0)
        _, pred = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (pred == labels).sum().item()

    train_acc = 100 * correct / total
    train_loss = running_loss / total
    val_acc = evaluate(model, val_loader)
    scheduler.step()

    print(f"Phase 1 - Epoch {epoch+1}: Train Loss: {train_loss:.4f}, "
          f"Train Acc: {train_acc:.2f}%, Val Acc: {val_acc:.2f}%")

    if val_acc > best_acc:
        best_acc = val_acc
        torch.save(model.state_dict(), '/content/drive/MyDrive/best_phase1_weights.pth')
        print(f"New best saved: {best_acc:.2f}%")

print(f"Phase 1 Best Val Acc: {best_acc:.2f}%")
```

```python
print("\nPHASE 2: Unfreezing blocks[6] + head...")

for name, param in model.named_parameters():
    if ('blocks.6' in name or 'conv_head' in name or
        'bn2' in name or 'classifier' in name):
        param.requires_grad = True
    else:
        param.requires_grad = False

freeze_bn(model)

optimizer = optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4, weight_decay=1e-3
)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=6)

patience = 2
patience_counter = 0

for epoch in range(6):
    model.train()
    freeze_bn(model)
    correct, total, running_loss = 0, 0, 0
    for images, labels in tqdm(train_loader, desc=f"Phase 2 - Epoch {epoch+1}"):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * labels.size(0)
        _, pred = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (pred == labels).sum().item()

    train_acc = 100 * correct / total
    train_loss = running_loss / total
    val_acc = evaluate(model, val_loader)
    scheduler.step()

    print(f"Phase 2 - Epoch {epoch+1}: Train Loss: {train_loss:.4f}, "
          f"Train Acc: {train_acc:.2f}%, Val Acc: {val_acc:.2f}%")

    if val_acc > best_acc:
        best_acc = val_acc
        patience_counter = 0
        torch.save(model.state_dict(), '/content/drive/MyDrive/best_efficientnet_b3.pth')
        print(f"\nNew best saved: {best_acc:.2f}%")
    else:
        patience_counter += 1
        print(f"\nNo improvement ({patience_counter}/{patience})")
        if patience_counter >= patience:
            print(f"Early stopping di epoch {epoch+1}")
            break

print(f"Best Val Acc after Phase 2: {best_acc:.2f}%")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix

# Create Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
print(f"Confusion matrix shape: {cm.shape}")

# Find Top Confused Pairs
cm_no_diag = cm.copy()
np.fill_diagonal(cm_no_diag, 0)

confused_pairs = []

for i in range(len(test_dataset.classes)):
    for j in range(len(test_dataset.classes)):
        if i != j and cm_no_diag[i, j] > 0:
            confused_pairs.append((cm_no_diag[i, j], test_dataset.classes[i], test_dataset.classes[j]))

confused_pairs.sort(reverse=True)

print("\nTop 20 Most Confused Pairs (Test Set):")
print(f"{'Count':<8} {'True Class':<25} {'Predicted As':<25}")
print("-" * 60)
for count, true_cls, pred_cls in confused_pairs[:20]:
    print(f"{count:<8} {true_cls:<25} {pred_cls:<25}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

top_classes = set()
for _, true_cls, pred_cls in confused_pairs[:20]:
    top_classes.add(true_cls)
    top_classes.add(pred_cls)

top_classes = sorted(top_classes)
top_indices = [test_dataset.classes.index(c) for c in top_classes]

# Subset confusion matrix
cm_subset = cm[np.ix_(top_indices, top_indices)]

plt.figure(figsize=(15, 12))
sns.heatmap(cm_subset, cmap='Blues', annot=True, fmt='d', cbar=True,
            xticklabels=top_classes, yticklabels=top_classes)
plt.title('Confusion Matrix — Top Confused Classes', fontsize=14)
plt.xlabel('Predicted', fontsize=12)
plt.ylabel('True', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/confusion_matrix_top20.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved to /content/drive/MyDrive/confusion_matrix_top20.png")
```
